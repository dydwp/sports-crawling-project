```python
import os
import pandas as pd
import mysql.connector

from dotenv import load_dotenv
```

## 전처리 데이터 불러오기

02단계에서 전처리한 CSV 파일을 불러온다.

```python
# 전처리 데이터 불러오기
input_file = 'sports_crawling_raw.csv'

df = pd.read_csv(
    input_file,
    encoding='utf-8-sig'
)

print('데이터 불러오기 완료')
print('데이터 크기:', df.shape)

display(df)
```

## MySQL 서버 연결

.env 파일에 저장된 데이터베이스 접속 정보를 이용하여
MySQL 서버에 연결한다.

```python
# MySQL 서버 연결
conn = mysql.connector.connect(
    host=DB_HOST,
    user=DB_USER,
    password=DB_PASSWORD
)

cursor = conn.cursor()

print('MySQL 연결 성공')
```

## 데이터베이스 생성

체육시설 크롤링 데이터를 저장하기 위한
데이터베이스를 생성한다.

데이터베이스 이름은 .env 파일의 DB_NAME 값을 사용한다.

```python
# 데이터베이스 생성
cursor.execute(
    f'CREATE DATABASE IF NOT EXISTS {DB_NAME}'
)

print('데이터베이스 생성 완료:', DB_NAME)
```

```python
# 사용할 데이터베이스 선택
conn.database = DB_NAME

print('사용 데이터베이스:', DB_NAME)
```

## 체육시설 테이블 생성

전처리된 체육시설 데이터를 저장하기 위한
sports_facility 테이블을 생성한다.

서비스ID는 각 체육시설 서비스를 구분하는
고유 식별자로 사용한다.

```python
# 테이블 생성

table_name = 'sports_facility'

create_table_sql = f'''
CREATE TABLE IF NOT EXISTS {table_name} (
    서비스구분 VARCHAR(20),
    서비스ID VARCHAR(30) PRIMARY KEY,
    대분류명 VARCHAR(50),
    소분류명 VARCHAR(50),
    서비스상태 VARCHAR(20),
    서비스명 VARCHAR(500)
)
'''

cursor.execute(create_table_sql)

print('테이블 생성 완료:', table_name)
```

## 전처리 데이터 불러오기

02단계에서 생성한 전처리 완료 데이터를
Pandas DataFrame으로 불러온다.

```python
# 전처리 데이터 불러오기
input_file = 'sports_crawling_raw.csv'

df = pd.read_csv(
    input_file,
    encoding='utf-8-sig'
)

print('전처리 데이터 불러오기 완료')
print('데이터 크기:', df.shape)

display(df)
```

## 체육시설 데이터 저장

Pandas DataFrame에 저장된 전처리 데이터를
MySQL의 sports_facility 테이블에 저장한다.

```python
# 데이터 저장 SQL
insert_sql = f'''
INSERT INTO {table_name}
(
    서비스구분,
    서비스ID,
    대분류명,
    소분류명,
    서비스상태,
    서비스명
)
VALUES (%s, %s, %s, %s, %s, %s)
'''
```

```python
# DataFrame 데이터를 MySQL에 저장
data_to_insert = [
    tuple(row)
    for row in df.itertuples(
        index=False,
        name=None
    )
]

cursor.executemany(
    insert_sql,
    data_to_insert
)

conn.commit()

print('데이터 저장 완료')
print('저장 데이터:', cursor.rowcount, '건')
```

## 데이터 저장 결과 확인

MySQL 테이블에 저장된 데이터의 개수를 조회하여
정상적으로 저장되었는지 확인한다.

```python
# 저장된 데이터 개수 확인
cursor.execute(
    f'SELECT COUNT(*) FROM {table_name}'
)

count = cursor.fetchone()[0]

print('MySQL 저장 데이터:', count, '건')
```

## 데이터 조회 및 검증

MySQL에 저장된 체육시설 데이터를 SELECT 문으로 조회하고
Pandas DataFrame으로 변환하여 확인한다.

```python
# MySQL 데이터 조회

select_sql = f'''
SELECT
    서비스구분,
    서비스ID,
    대분류명,
    소분류명,
    서비스상태,
    서비스명
FROM {table_name}
'''

cursor.execute(select_sql)

rows = cursor.fetchall()

db_df = pd.DataFrame(
    rows,
    columns=[
        '서비스구분',
        '서비스ID',
        '대분류명',
        '소분류명',
        '서비스상태',
        '서비스명'
    ]
)

display(db_df)
```

```python
# CSV와 MySQL 데이터 개수 비교
csv_count = len(df)
db_count = len(db_df)

print('CSV 데이터:', csv_count, '건')
print('MySQL 데이터:', db_count, '건')

if csv_count == db_count:
    print('데이터 저장 검증 성공')
else:
    print('데이터 저장 검증 실패')
```

## MySQL 연결 종료

데이터 저장 및 검증이 완료되었으므로
MySQL 연결을 종료한다.

```python
# MySQL 연결 종료
cursor.close()
conn.close()

print('MySQL 연결 종료 완료')
```
