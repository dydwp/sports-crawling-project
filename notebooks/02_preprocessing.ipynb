# 체육시설 데이터 전처리

## 프로젝트 개요

01단계에서 Selenium을 이용하여 수집한
서울시 체육시설 공공서비스예약 데이터를 불러와
Pandas를 이용하여 전처리를 수행한다.

### 전처리 과정

1. 원본 데이터 불러오기
2. 데이터 구조 및 기본 정보 확인
3. 결측치 확인
4. 중복 데이터 확인 및 제거
5. 문자열 데이터 정리
6. 전처리 결과 확인
7. 최종 데이터 저장

```python
import pandas as pd
```

```python
# 원본 크롤링 데이터 불러오기
input_file = 'sports_crawling_raw.csv'

df = pd.read_csv(
    input_file,
    encoding='utf-8-sig'
)

print('원본 데이터 불러오기 완료')
print('데이터 크기:', df.shape)
```

## 데이터 기본 정보 확인

크롤링한 원본 데이터의 행과 열,
컬럼명 및 데이터 타입을 확인한다.

```python
# 데이터 앞부분 확인
display(df.head())
```

```python
# 데이터 크기 확인
print('행 개수:', df.shape[0])
print('열 개수:', df.shape[1])
```

```python
# 컬럼명 확인
print('컬럼 목록')

for i, column in enumerate(df.columns):
    print(f"{i}. {column}")
```

```python
# 데이터 타입 확인
df.info()
```

## 결측치 확인 및 처리

각 컬럼별 결측치 개수를 확인하고,
필요한 경우 결측 데이터를 처리한다.

```python
# 컬럼별 결측치 확인
missing = df.isnull().sum()

print('컬럼별 결측치')
display(missing.to_frame('결측치 개수'))
```

```python
# 전체 결측치 개수
print('전체 결측치:', df.isnull().sum().sum())
```

## 중복 데이터 확인 및 제거

동일한 데이터가 여러 번 수집되었는지 확인하고,
중복 데이터가 존재할 경우 제거한다.

서비스ID를 기준으로 데이터의 고유성을 확인한다.

```python
# 전체 행 기준 중복 확인
duplicate_count = df.duplicated().sum()

print('전체 행 중복 데이터:', duplicate_count, '건')
```

```python
# 서비스ID 중복 확인
service_id_duplicate = df['서비스ID'].duplicated().sum()

print('서비스ID 중복 데이터:', service_id_duplicate, '건')
```

```python
# 서비스ID 기준 중복 제거
df = df.drop_duplicates(
    subset=['서비스ID'],
    keep='first'
)

print('중복 제거 후 데이터:', len(df), '건')
```

## 문자열 데이터 정리

서비스명 및 분류 정보에 포함될 수 있는
불필요한 앞뒤 공백을 제거한다.

```python
# 문자열 컬럼의 앞뒤 공백 제거
text_columns = [
    '서비스구분',
    '서비스ID',
    '대분류명',
    '소분류명',
    '서비스상태',
    '서비스명'
]

for column in text_columns:
    df[column] = df[column].astype(str).str.strip()

print('문자열 데이터 정리 완료')
```

```python
# 빈 문자열 확인
for column in text_columns:
    empty_count = (df[column] == '').sum()
    
    if empty_count > 0:
        print(f'{column}: {empty_count}건')
```

```python
# 서비스 상태별 데이터 개수
status_count = df['서비스상태'].value_counts()

print('서비스 상태별 데이터 개수')
display(status_count)
```

## 전처리 결과 확인

모든 전처리 작업이 완료된 데이터를 확인한다.

```python
# 최종 데이터 확인
print('최종 데이터 크기:', df.shape)

display(df)
```

```python
# 최종 결측치 확인

print('최종 결측치:', df.isnull().sum().sum())
```

```python
# 최종 중복 확인
print('최종 중복 데이터:', df.duplicated().sum())
```

## 전처리 데이터 저장

전처리가 완료된 데이터를 CSV 파일로 저장한다.

저장된 파일은 다음 단계에서 MySQL 데이터베이스에
저장하기 위한 입력 데이터로 사용한다.

```python
# 전처리 완료 데이터 저장
output_file = 'sports_crawling_raw.csv'

df.to_csv(
    output_file,
    index=False,
    encoding='utf-8-sig'
)

print('전처리 데이터 저장 완료')
print('파일명:', output_file)
print('저장 데이터:', len(df), '건')
```
